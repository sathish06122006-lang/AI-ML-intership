```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_student_scores.csv")

print(df.head())
```

```python
plt.scatter(df["math_score"], df["physics_score"])

plt.xlabel("Math Score")
plt.ylabel("Physics Score")
plt.title("Math Score vs Physics Score")

plt.show()
```

```python
plt.plot(df["math_score"].head(10), marker="o", label="Math")
plt.plot(df["physics_score"].head(10), marker="o", label="Physics")
plt.plot(df["chemistry_score"].head(10), marker="o", label="Chemistry")

plt.xlabel("Student")
plt.ylabel("Score")
plt.title("Student Score Trend")

plt.legend()
plt.show()
```

```python
subjects = [
    "math_score",
    "history_score",
    "physics_score",
    "chemistry_score",
    "biology_score",
    "english_score",
    "geography_score"
]

average_scores = df[subjects].mean()

plt.bar(subjects, average_scores)

plt.xlabel("Subjects")
plt.ylabel("Average Score")
plt.title("Average Student Scores")

plt.xticks(rotation=45)
plt.show()
```
